```python

# Data luas bangunan (X) dan harga (Y)
x = [700, 800, 400, 300, 575, 300, 350, 300, 69, 42]
y = [28, 19, 4.7, 4.9, 28, 10, 7.6, 5.25, 0.67, 0.48]

# Jumlah data
m = len(x)

# Menghitung jumlah setiap kolom
sum_x = sum(x)
sum_y = sum(y)
sum_x2 = sum(i**2 for i in x)
sum_xy = sum(x[i] * y[i] for i in range(m))

print("Jumlah data =", m)
print("Jumlah X =", sum_x)
print("Jumlah Y =", sum_y)
print("Jumlah X² =", sum_x2)
print("Jumlah XY =", round(sum_xy, 2))

# Menghitung nilai A (slope)
a = (m * sum_xy - sum_x * sum_y) / (
    m * sum_x2 - sum_x**2
)

# Menghitung nilai B (intercept)
b = (sum_y - a * sum_x) / m

print("\nNilai A =", round(a, 5))
print("Nilai B =", round(b, 5))

# Prediksi harga untuk luas bangunan 500
luas = 500
harga = a * luas + b

print("\nLuas bangunan =", luas)
print("Prediksi harga =", round(harga, 2), "miliar")
```
